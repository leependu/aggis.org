### Exercise 3

You have been provided below with a function that use haversine formula to calculate the distance between two locations (expressed with latitude and longitude coordinates).  Based on what you know about how functions work, look over the function and figure out how to call the function.  Then, prompt the user to enter the latitude and longitude values for two locations, and calculate the distance between them by calling the function.

*Hint: Remember that typecasting strings to decimal numbers is accomplished using float().*

```python
from math import radians, cos, sin, asin, sqrt, degrees, pi, atan2

def haversine(lat1, lon1, lat2, lon2):
  lat1 = radians(lat1)
  lon1 = radians(lon1)
  lat2 = radians(lat2)
  lon2 = radians(lon2)

  d_lat = lat2 - lat1
  d_lon = lon2 - lon1
  d = sin(d_lat * 0.5) ** 2 + cos(lat1) * cos(lat2) * sin(d_lon * 0.5) ** 2
  return 12742.0176 * asin(sqrt(d)) # distance in kilometers

lat1 = float(input("Please enter the first point's Y coordinate in decimal degrees: ")) # user lat1 / y1 input
lon1 = float(input("Please enter the first point's X coordinate in decimal degrees: ")) # user lon1 / x1 input
lat2 = float(input("Please enter the second point's Y coordinate in decimal degrees: ")) # user lat2 / y2 input
lon2 = float(input("Please enter the second point's X coordinate in decimal degrees: ")) # user lon2 / x2 input
print("These two points are...")
print(haversine(lat1, lon1, lat2, lon2)) # distance in kilometers
print("km apart")
```
